```python
# PURCHASE_TIME ANALYSIS....
```

```python
data["PurchaseTime"] = pd.to_datetime(data["PurchaseTime"].astype(str))
```

```python
data["PurchaseTime"].dtype
```

```python
def get_time_period(hour):
    if 5 <= hour < 12:
        return "Morning"
    elif 12 <= hour < 17:
        return "Afternoon"
    elif 17 <= hour < 21:
        return "Evening"
    else:
        return "Night"

data["TimePeriod"] = data["PurchaseTime"].dt.hour.apply(get_time_period)

data[["PurchaseTime", "TimePeriod"]].head(11)
```

```python
time_sales = data.groupby('TimePeriod')["TotalAmount(INR)"].sum()
time_sales
```
